# CP Stats Sheet — Build

Concatenates all source files into a single self-contained HTML file.

**Workflow:**
1. Run Cell 1 to build `dist/cpstatssheet.html`
2. Run Cell 2 to open the built file in your browser
3. Run Cell 3 to check line counts across source files

No npm, no node, no external dependencies — just Python string concatenation.

```python
# ============================================================
# Cell 1: BUILD — assembles dist/cpstatssheet.html
# ============================================================
import os
import glob

src_dir  = os.path.join(os.path.dirname(os.path.abspath('__file__')), 'src')
dist_dir = os.path.join(os.path.dirname(os.path.abspath('__file__')), 'dist')
os.makedirs(dist_dir, exist_ok=True)

# Read CSS
with open(os.path.join(src_dir, '01_styles.css'), encoding='utf-8') as f:
    styles = f.read()

# Read shell.html (the static HTML skeleton)
with open(os.path.join(src_dir, 'shell.html'), encoding='utf-8') as f:
    shell = f.read()

# Read all JS files in order (02_ through 99_)
js_files = sorted(glob.glob(os.path.join(src_dir, '0[2-9]_*.js')) +
                  glob.glob(os.path.join(src_dir, '[1-9][0-9]_*.js')))

js_parts = []
for path in js_files:
    with open(path, encoding='utf-8') as f:
        js_parts.append(f'/* {os.path.basename(path)} */\n' + f.read())

main_js = '\n\n'.join(js_parts)

# Assemble the single-file HTML
html = f"""<!DOCTYPE html>
<html lang=\"en\">
<head>
  <meta charset=\"UTF-8\">
  <meta name=\"viewport\" content=\"width=device-width, initial-scale=1.0\">
  <title>CP Stats Sheet</title>
  <style>
{styles}
  </style>
{shell}
<script>
{main_js}
</script>
</body>
</html>"""

out_path = os.path.join(dist_dir, 'cpstatssheet.html')
with open(out_path, 'w', encoding='utf-8') as f:
    f.write(html)

size_kb = os.path.getsize(out_path) / 1024
print(f'Built: {out_path}')
print(f'Size:  {size_kb:.1f} KB')
print(f'JS files included: {len(js_files)}')
for p in js_files:
    print(f'  {os.path.basename(p)}')
```

```python
# ============================================================
# Cell 2: OPEN in browser
# ============================================================
import webbrowser, os
out_path = os.path.join(os.path.dirname(os.path.abspath('__file__')), 'dist', 'cpstatssheet.html')
webbrowser.open('file://' + os.path.abspath(out_path))
print('Opened in browser.')
```

```python
# ============================================================
# Cell 3: LINE COUNT report
# ============================================================
import os, glob
src_dir = os.path.join(os.path.dirname(os.path.abspath('__file__')), 'src')
files   = sorted(glob.glob(os.path.join(src_dir, '*')))
total   = 0
print(f"{'File':<35} {'Lines':>7}")
print('-' * 44)
for p in files:
    with open(p, encoding='utf-8') as f:
        n = sum(1 for _ in f)
    total += n
    print(f"{os.path.basename(p):<35} {n:>7,}")
print('-' * 44)
print(f"{'TOTAL':<35} {total:>7,}")
```
