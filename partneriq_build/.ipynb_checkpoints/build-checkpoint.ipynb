# Workflow 1 — Development
**Run Cell 1, then Cell 2, then Cell 3 — in that order.**

````python
# ============================================================
# CELL 1 — BUILD
# Run this cell to assemble dist/partneriq.html
# ============================================================
import os, glob, subprocess
from pathlib import Path

SRC  = Path('src')
DIST = Path('dist')
OUT  = DIST / 'partneriq.html'
DIST.mkdir(exist_ok=True)

# Read a source file
def read(filename):
    return (SRC / filename).read_text(encoding='utf-8')

# Collect all JS files in order (00_ through 15_)
js_files = sorted(SRC.glob('[0-9][0-9]_*.js'))

vendor_shim  = (SRC / '00_vendor_shims.js').read_text(encoding='utf-8')
styles       = (SRC / '01_styles.css').read_text(encoding='utf-8')
shell        = (SRC / 'shell.html').read_text(encoding='utf-8')

# Main JS = everything from 02 onwards
main_js_files = sorted(f for f in js_files if not f.name.startswith('00_'))
main_js = '\n\n'.join(f.read_text(encoding='utf-8') for f in main_js_files)

# Load partner logos from logos/ folder and encode for embedding
import json as _json, base64 as _base64

def _build_partner_logos_js():
    logos_dir = Path('logos')
    if not logos_dir.exists():
        print('  ⚠ logos/ folder not found — no partner logos will appear')
        return 'const PARTNER_LOGOS = {}; // logos/ folder not found'
    logo_files = sorted(f for pat in ('*.png','*.jpg','*.jpeg','*.svg') for f in logos_dir.glob(pat))
    if not logo_files:
        print('  ⚠ logos/ is empty — no partner logos will appear')
        return 'const PARTNER_LOGOS = {}; // no logos loaded'
    entries = {}
    for lf in logo_files:
        ext = lf.suffix.lower()
        if ext == '.svg':
            entries[lf.stem] = {'type': 'svg', 'data': lf.read_text(encoding='utf-8')}
        else:
            mime = 'image/jpeg' if ext in ('.jpg', '.jpeg') else 'image/png'
            b64 = _base64.b64encode(lf.read_bytes()).decode('ascii')
            entries[lf.stem] = {'type': 'img', 'data': f'data:{mime};base64,{b64}'}
        print(f'  ✓ Logo loaded: {lf.stem!r}  ({lf.stat().st_size // 1024} KB  {lf.name})')
    return f'const PARTNER_LOGOS = {_json.dumps(entries, ensure_ascii=False)};'

partner_logos_js = _build_partner_logos_js()

html = f"""<!DOCTYPE html>
<html lang=\"en\">
<head>
<meta charset=\"UTF-8\" />
<meta name=\"viewport\" content=\"width=device-width, initial-scale=1.0\" />
<title>PartnerIQ · Brand Performance Intelligence</title>
<script>
{vendor_shim}
</script>
<style>
{styles}
</style>
</head>
{shell}
<script>
{partner_logos_js}
</script>
<script>
{main_js}
</script>
</body>
</html>
"""

OUT.write_text(html, encoding='utf-8')

# Report
size_kb = OUT.stat().st_size // 1024
line_count = html.count('\n')
print(f'✓ Built: {OUT}')
print(f'  Size:  {size_kb} KB')
print(f'  Lines: {line_count:,}')
print(f'  JS files assembled: {len(main_js_files)}')
for f in main_js_files:
    lc = f.read_text(encoding='utf-8').count('\n')
    print(f'    {f.name:40s}  {lc:4d} lines')
````

````python
# ============================================================
# CELL 2 — VALIDATE (Workflow 1, step 2)
# Checks the built file for JavaScript syntax errors.
# Requires Node.js. If not installed, skip this cell.
# ============================================================
import subprocess, re
from pathlib import Path

html = Path('dist/partneriq.html').read_text(encoding='utf-8')

# Extract the main script block only (skip the vendor shim which contains non-JS content)
scripts = re.findall(r'<script[^>]*>(.*?)</script>', html, re.DOTALL)
main_js = max(scripts, key=len)

try:
    result = subprocess.run(
        ['node', '--check'],
        input=main_js, capture_output=True, text=True, timeout=15
    )
    if result.returncode == 0:
        print('✓ JavaScript syntax: clean')
    else:
        print('✗ Syntax error found:')
        print(result.stderr)
        m = re.search(r':(\d+)', result.stderr)
        if m:
            ln = int(m.group(1))
            lines = main_js.split('\n')
            for i in range(max(0, ln-3), min(len(lines), ln+2)):
                print(f'  {i+1:5d}: {lines[i]}')
except FileNotFoundError:
    print('⚠ Node.js not found — skipping syntax check.')
    print('  The file should still work fine in the browser.')
````

# Workflow 2 — AM Export
**Before running these cells**, complete the manual steps:
1. Open `dist/partneriq.html` in your browser
2. Load all data files via Import / Export
3. Click **Export preloaded dashboard** inside the dashboard
4. Move the exported file into this folder (`partneriq_build/`)
5. Rename it to exactly `partneriq_preloaded.html`

**Then run Cell 4, then Cell 5 — in that order.**

````python
# ============================================================
# CELL 4 — CREATE MINIFIED AM EXPORT (Workflow 2, step 1)
# ============================================================
# Order of operations:
#   1. Load data into dist/partneriq.html (open it in browser, drag in your files)
#   2. Click 'Export preloaded dashboard' inside the dashboard
#   3. Move that exported file into this folder and rename it partneriq_preloaded.html
#   4. Run THIS cell — it minifies the code and produces dist/partneriq_export.html
#   5. Send dist/partneriq_export.html to AMs
#
# The export has: all data baked in, viewer mode on, import/export hidden,
#                 code minified so source is not human-readable.
# ============================================================

# Install minification libraries if not already present (runs once)
import subprocess, sys
for pkg in ['rjsmin', 'rcssmin']:
    try:
        __import__(pkg)
    except ImportError:
        print(f'Installing {pkg}...')
        subprocess.check_call([sys.executable, '-m', 'pip', 'install', pkg, '-q'])

import re, rjsmin, rcssmin
from pathlib import Path
from datetime import date

# ── Config ──────────────────────────────────────────────────
VERSION   = 'v0.22'              # update each release
RECIPIENT = 'AM'                 # optional label in filename
SOURCE    = Path('partneriq_preloaded.html')   # drop your export here
OUT_DIR   = Path('dist')
# ────────────────────────────────────────────────────────────

if not SOURCE.exists():
    print(f'ERROR: {SOURCE} not found.')
    print('Steps:')
    print('  1. Open dist/partneriq.html in your browser')
    print('  2. Load your data files via Import/Export')
    print('  3. Click "Export preloaded dashboard" in the dashboard')
    print(f'  4. Move that file here and rename it to {SOURCE}')
    print('  5. Run this cell again')
else:
    html = SOURCE.read_text(encoding='utf-8')
    original_size = len(html.encode())

    # Minify all <script> blocks
    def minify_script(match):
        tag   = match.group(1)   # any attributes on the tag
        code  = match.group(2)
        try:
            minified = rjsmin.jsmin(code, keep_bang_comments=False)
        except Exception:
            minified = code      # fall back to original if minifier chokes
        return f'<script{tag}>{minified}</script>'

    # Minify <style> blocks
    def minify_style(match):
        tag  = match.group(1)
        code = match.group(2)
        try:
            minified = rcssmin.cssmin(code)
        except Exception:
            minified = code
        return f'<style{tag}>{minified}</style>'

    html_min = re.sub(r'<script([^>]*)>(.*?)</script>', minify_script, html, flags=re.DOTALL)
    html_min = re.sub(r'<style([^>]*)>(.*?)</style>',   minify_style,  html_min, flags=re.DOTALL)

    # Remove HTML comments (except IE conditionals)
    html_min = re.sub(r'<!--(?!\[if).*?-->', '', html_min, flags=re.DOTALL)

    # Write output
    today    = date.today().strftime('%Y-%m-%d')
    out_name = f'partneriq_export_{VERSION}_{today}.html'
    out_path = OUT_DIR / out_name
    out_path.write_text(html_min, encoding='utf-8')

    final_size  = len(html_min.encode())
    reduction   = (1 - final_size / original_size) * 100

    print(f'✓  Export created: {out_path}')
    print(f'   Original:  {original_size // 1024:,} KB')
    print(f'   Minified:  {final_size // 1024:,} KB  ({reduction:.0f}% smaller)')
    print()
    print('This file is ready to send to account managers.')
    print('Code is minified — source is not human-readable.')
    print('Data is embedded — no internet connection required to view.')
````

````python
# ============================================================
# CELL 5 — COPY TO DOWNLOADS (Workflow 2, step 2)
# Copies the minified export to your Downloads folder.
# Run AFTER Cell 4.
# ============================================================
import shutil
from pathlib import Path
from datetime import date

VERSION = 'v0.23'  # ← update this each release

# Find the latest export file produced by Cell 4
export_files = sorted(Path('dist').glob('partneriq_export_*.html'))
if not export_files:
    print('No export file found in dist/.')
    print('Run Cell 4 first to create the minified export.')
else:
    src  = export_files[-1]  # most recently created export
    today = date.today().strftime('%Y-%m-%d')
    dest = Path.home() / 'Downloads' / f'partneriq_{VERSION}_{today}.html'
    shutil.copy2(src, dest)
    print(f'✓ Copied to: {dest}')
    print(f'  Source: {src.name} ({src.stat().st_size // 1024} KB)')
    print()
    print('This is the file to send to account managers.')
````

# Workflow 3 — Utilities
Run these individually as needed. They are not part of any sequence.

````python
# ============================================================
# CELL 6 — LINE COUNT REPORT (Utility)
# Shows how many lines are in each source file.
# Useful for spotting which files are getting too large.
# ============================================================
from pathlib import Path

src = Path('src')
files = sorted(src.glob('*.js')) + sorted(src.glob('*.css')) + sorted(src.glob('*.html'))

total = 0
print(f'{"File":<45} {"Lines":>6}  {"KB":>5}')
print('-' * 60)
for f in files:
    text = f.read_text(encoding='utf-8')
    lc = text.count('\n') + 1
    kb = len(text.encode()) // 1024
    total += lc
    print(f'{f.name:<45} {lc:>6}  {kb:>4}KB')
print('-' * 60)
print(f'{"TOTAL":<45} {total:>6}')
````

````python
# ============================================================
# CELL 7 — LOGO FOLDER CHECKER (Utility)
# Scans the logos/ folder and checks which partners have logos
# and which don't. Helps you spot naming mismatches before building.
# ============================================================
from pathlib import Path

LOGOS_DIR = Path('logos')

if not LOGOS_DIR.exists():
    print('No logos/ folder found.')
    print(f'Create it at: {Path.cwd() / "logos"}')
else:
    logo_files = sorted(
        list(LOGOS_DIR.glob('*.png')) +
        list(LOGOS_DIR.glob('*.jpg')) +
        list(LOGOS_DIR.glob('*.jpeg')) +
        list(LOGOS_DIR.glob('*.svg'))
    )

    if not logo_files:
        print('logos/ folder is empty.')
        print('Add PNG files named after your partners, e.g. "Alaska Airlines.png"')
    else:
        import base64
        print(f'Found {len(logo_files)} logo file(s):\n')
        total_kb = 0
        print(f'{"Partner name (filename)":<40} {"Format":<6} {"Size":>8}')
        print('-' * 58)
        for lf in logo_files:
            kb = lf.stat().st_size // 1024
            b64_kb = len(base64.b64encode(lf.read_bytes())) // 1024 if lf.suffix != '.svg' else kb
            total_kb += b64_kb
            fmt = lf.suffix.upper().lstrip('.')
            print(f'{lf.stem:<40} {fmt:<6} {kb:>5} KB  (~{b64_kb} KB base64)')
        print('-' * 58)
        print(f'{"TOTAL base64 size added to build":>50} ~{total_kb} KB')
        print()
        print('Naming tips:')
        print('  • Filename must match canonical partner name exactly (case-sensitive)')
        print('  • Check the partner name in the PartnerIQ dashboard search bar')
        print('  • Run Cell 1 after adding logos — mismatches are reported as warnings')
        print()
        print('Org logo (Blazers) is set separately in src/16_report_template.js')
````
